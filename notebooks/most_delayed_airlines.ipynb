```python
import sys
sys.path.append('/Users/nf/Documents/Projects/SkyLake')

from spark import config

session = config.Session()
spark = session.get_spark_session()
```

```python
# "Which airline delays most?"

spark.sql('''
    SELECT
        OP_CARRIER,
        ROUND(AVG(ARR_DELAY), 2) AS avg_delay,
        COUNT(*) AS total_flights
    FROM local.flights.flight_events
    WHERE ARR_DELAY > 0
    GROUP BY OP_CARRIER
    ORDER BY avg_delay DESC
    LIMIT 10
''').show()
```
